```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = None   # the dataset will be stored here after loading

print("Welcome to the World Happiness Analysis Program!")

while True:
    print("\nPlease select an option from the menu below:")
    print("1. Load the dataset")
    print("2. Explore the dataset")
    print("3. Clean the dataset")
    print("4. Analyze the dataset")
    print("5. Correlation analysis")
    print("6. Visualize the dataset")
    print("7. Exit")
    choice = input("Enter your choice from 1 to 7: ")

    # ------------------------- 1. LOAD -------------------------
    if choice == "1":
        file_path = input("Enter the path of the file (CSV file): ")
        try:
            data = pd.read_csv(file_path)
            data.columns = data.columns.str.strip()   # remove extra spaces from column names
            print("Dataset loaded successfully!")
        except Exception as e:
            print("Error loading file:", e)

    # ------------------------- 7. EXIT -------------------------
    elif choice == "7":
        print("Thank you for using the program. Goodbye!")
        break

    elif choice not in ["2", "3", "4", "5", "6"]:
        print("Invalid choice. Please enter a number between 1 and 7.")

    elif data is None:
        print("Load the dataset first by selecting the first choice in the menu")

    # ------------------------- 2. EXPLORE -------------------------
    elif choice == "2":
        while True:
            print("\n--- Explore the dataset ---")
            print("1. First 5 rows")
            print("2. Number of rows and columns")
            print("3. Column names and data types")
            print("4. Summary statistics")
            print("5. Missing values in each column")
            print("6. Back to the main menu")
            sub = input("Enter your choice from 1 to 6: ")

            if sub == "1":
                print(data.head())
            elif sub == "2":
                print("Rows:", data.shape[0], " Columns:", data.shape[1])
            elif sub == "3":
                print(data.dtypes)
            elif sub == "4":
                print(data.describe())
            elif sub == "5":
                print(data.isnull().sum())
            elif sub == "6":
                break
            else:
                print("Invalid choice, try again.")

    # ------------------------- 3. CLEAN -------------------------
    elif choice == "3":
        while True:
            print("\n--- Clean the dataset ---")
            print("Missing values in each column:")
            print(data.isnull().sum())
            print("1. Drop rows with missing values")
            print("2. Fill with mean")
            print("3. Fill with median")
            print("4. Fill with mode")
            print("5. Remove duplicates")
            print("6. Back to the main menu")
            sub = input("Enter your choice from 1 to 6: ")

            if sub == "1":
                data.dropna(inplace=True)
                print("Missing values dropped.")
            elif sub == "2" or sub == "3":
                num_cols = data.select_dtypes(include="number").columns
                for col in num_cols:
                    fill_val = data[col].mean() if sub == "2" else data[col].median()
                    data[col] = data[col].fillna(fill_val)
                # text columns cannot use mean/median, so fill them with mode
                text_cols = data.select_dtypes(exclude="number").columns
                for col in text_cols:
                    mode = data[col].mode()
                    if not mode.empty:
                        data[col] = data[col].fillna(mode.iloc[0])
                print("Missing values filled with", "mean." if sub == "2" else "median.")
            elif sub == "4":
                for col in data.columns:
                    mode = data[col].mode()
                    if not mode.empty:
                        data[col] = data[col].fillna(mode.iloc[0])
                print("Missing values filled with mode values.")
            elif sub == "5":
                before = len(data)
                data.drop_duplicates(inplace=True)
                data.reset_index(drop=True, inplace=True)
                print("Duplicate rows removed:", before - len(data))
            elif sub == "6":
                break
            else:
                print("Invalid choice, try again.")

    # ------------------------- 4. ANALYZE -------------------------
    elif choice == "4":
        while True:
            print("\n--- Analyze the dataset ---")
            print("1. Identify the happiest countries")
            print("2. Identify the least happy countries")
            print("3. GDP per capita")
            print("4. Social support")
            print("5. Life expectancy")
            print("6. Back to the main menu")
            sub = input("Enter your choice from 1 to 6: ")

            if sub == "1":
                print("Happiest countries")
                happiest = data.sort_values(by="Happiness Score", ascending=False).head()
                print(happiest[["Country", "Happiness Score"]])
            elif sub == "2":
                print("Least happy countries")
                least_happy = data.sort_values(by="Happiness Score", ascending=True).head()
                print(least_happy[["Country", "Happiness Score"]])
            elif sub == "3":
                col = "Economy (GDP per Capita)"
                print("Highest GDP per capita\n")
                print(data.sort_values(by=col, ascending=False).head()[["Country", "Happiness Score", col]])
                print("\nLowest GDP per capita\n")
                print(data.sort_values(by=col, ascending=True).head()[["Country", "Happiness Score", col]])
            elif sub == "4":
                col = "Family"
                print("Highest social support\n")
                print(data.sort_values(by=col, ascending=False).head()[["Country", "Happiness Score", col]])
                print("\nLowest social support\n")
                print(data.sort_values(by=col, ascending=True).head()[["Country", "Happiness Score", col]])
            elif sub == "5":
                col = "Health (Life Expectancy)"
                print("Highest life expectancy\n")
                print(data.sort_values(by=col, ascending=False).head()[["Country", "Happiness Score", col]])
                print("\nLowest life expectancy\n")
                print(data.sort_values(by=col, ascending=True).head()[["Country", "Happiness Score", col]])
            elif sub == "6":
                break
            else:
                print("Invalid choice, try again.")

    # ------------------------- 5. CORRELATION -------------------------
    elif choice == "5":
        while True:
            print("\n--- Correlation analysis ---")
            print("1. Correlation between happiness score and GDP per capita")
            print("2. Correlation between happiness score and social support")
            print("3. Correlation between happiness score and life expectancy")
            print("4. Back to the main menu")
            sub = input("Enter your choice from 1 to 4: ")

            if sub == "1":
                corr = data["Happiness Score"].corr(data["Economy (GDP per Capita)"])
                print("Correlation between happiness score and GDP per capita:", round(corr, 3))
            elif sub == "2":
                corr = data["Happiness Score"].corr(data["Family"])
                print("Correlation between happiness score and social support:", round(corr, 3))
            elif sub == "3":
                corr = data["Happiness Score"].corr(data["Health (Life Expectancy)"])
                print("Correlation between happiness score and life expectancy:", round(corr, 3))
            elif sub == "4":
                break
            else:
                print("Invalid choice, try again.")

    # ------------------------- 6. VISUALIZE -------------------------
    elif choice == "6":
        while True:
            print("\n--- Visualize the dataset ---")
            print("1. Bar chart of happiest countries")
            print("2. Bar chart of least happy countries")
            print("3. Scatter plot of Happiness Score vs GDP per Capita")
            print("4. Scatter plot of Happiness Score vs Social Support")
            print("5. Scatter plot of Happiness Score vs Life Expectancy")
            print("6. Heatmap of correlation between variables")
            print("7. Back to the main menu")
            sub = input("Enter your choice from 1 to 7: ")

            if sub == "1":
                top10 = data.sort_values(by="Happiness Score", ascending=False).head(10)
                plt.figure(figsize=(10, 6))
                sns.barplot(data=top10, x="Country", y="Happiness Score")
                plt.title("Happiest countries")
                plt.xlabel("Country")
                plt.ylabel("Happiness Score")
                plt.xticks(rotation=45)
                plt.tight_layout()
                plt.show()
            elif sub == "2":
                bottom10 = data.sort_values(by="Happiness Score", ascending=True).head(10)
                plt.figure(figsize=(10, 6))
                sns.barplot(data=bottom10, x="Country", y="Happiness Score")
                plt.title("Least happy countries")
                plt.xlabel("Country")
                plt.ylabel("Happiness Score")
                plt.xticks(rotation=45)
                plt.tight_layout()
                plt.show()
            elif sub == "3":
                plt.scatter(data["Economy (GDP per Capita)"], data["Happiness Score"])
                plt.title("Happiness Score vs GDP per Capita")
                plt.xlabel("GDP per Capita")
                plt.ylabel("Happiness Score")
                plt.show()
            elif sub == "4":
                plt.scatter(data["Family"], data["Happiness Score"])
                plt.title("Happiness Score vs Social Support")
                plt.xlabel("Social Support")
                plt.ylabel("Happiness Score")
                plt.show()
            elif sub == "5":
                plt.scatter(data["Health (Life Expectancy)"], data["Happiness Score"])
                plt.title("Happiness Score vs Life Expectancy")
                plt.xlabel("Life Expectancy")
                plt.ylabel("Happiness Score")
                plt.show()
            elif sub == "6":
                corr_matrix = data.corr(numeric_only=True)
                plt.figure(figsize=(9, 7))
                sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
                plt.title("Correlation Heatmap")
                plt.tight_layout()
                plt.show()
            elif sub == "7":
                break
            else:
                print("Invalid choice, try again.")
```
